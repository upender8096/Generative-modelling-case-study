**3.3 Correlation Heatmap**

```python
# Select numeric columns and ignore the ID column so it doesn't skew the heatmap
numeric_data = train.select_dtypes(include='number').drop(columns=['respondent_id'], errors='ignore')

# Calculate the correlation matrix
correlation_matrix = numeric_data.corr()

fig, ax = plt.subplots(figsize=(11, 9))

# Plot the heatmap (blue to red scale from -1 to 1)
im = ax.imshow(correlation_matrix, cmap='coolwarm', vmin=-1, vmax=1)

# Set the ticks to match the number of columns
ax.set_xticks(range(len(correlation_matrix)))
ax.set_yticks(range(len(correlation_matrix)))

# Add the column names as labels, rotating the x-axis ones to fit
ax.set_xticklabels(correlation_matrix.columns, rotation=90, fontsize=7)
ax.set_yticklabels(correlation_matrix.columns, fontsize=7)

# Add a colorbar legend to the side
fig.colorbar(im, fraction=0.046)

# Add title and adjust layout so labels don't get cut off
plt.title('Correlation Heatmap (Numeric Features)')
plt.tight_layout()

# Display the plot
plt.show()
```

**3.4 Strongest Predictors of Vaccination**

```python
# Define our target variable
target_col = 'covid_vaccine'

# Get correlations just for our target, and drop the target itself (to hide the 1.0 self-correlation)
top_correlations = correlation_matrix[target_col].drop(target_col)

# Sort by absolute value to find the strongest positive AND negative relationships
# Then, keep only the top 12 most important features
top_correlations = top_correlations.sort_values(key=abs, ascending=False).head(12)

# Plot as a horizontal bar chart (sort normally here so the largest bar is at the top of the graph)
top_correlations.sort_values().plot(kind='barh', color='#228833')

# Add a title
plt.title('Top correlations with covid_vaccine')
plt.show()

# Print the actual correlation values rounded to 3 decimal places
print(top_correlations.round(3))
```

3.5 Opinion & Behaviour vs Vaccination

```python
# Make a copy to avoid modifying the original dataset
eda_data = train.copy()

# Group all the behavioral precaution columns into a list
behavior_cols = [
    'behavioral_antiviral_meds', 'behavioral_avoidance', 'behavioral_face_mask',
    'behavioral_wash_hands', 'behavioral_large_gatherings',
    'behavioral_outside_home', 'behavioral_touch_face'
]

# Create a new feature calculating the total number of precautions each person took
eda_data['total_precautions'] = eda_data[behavior_cols].sum(axis=1)

# Set up 3 side-by-side subplots
fig, axes = plt.subplots(1, 3, figsize=(15, 4))

# Define the columns we want to plot and their readable titles
columns_to_plot = ['opinion_covid_risk', 'opinion_covid_vacc_effective', 'total_precautions']
chart_titles = ['Perceived Risk', 'Believed Effectiveness', 'Total Precautions']
target_col = 'covid_vaccine'

# Loop through each feature and plot the average vaccination rate
for ax, col, title in zip(axes, columns_to_plot, chart_titles):
    eda_data.groupby(col)[target_col].mean().plot(kind='bar', ax=ax, color='#4477aa')
    ax.set_title('Vaccination Rate by ' + title)
    ax.set_ylabel('Rate')

# Adjust layout so the titles and labels do not overlap
plt.tight_layout()
plt.show()
```

```python
# Drop the target column and fill any missing values with the median
numeric_features = numeric_data.drop(columns=[target_col], errors='ignore')
numeric_features_clean = numeric_features.fillna(numeric_features.median())

# Scale the data, then apply PCA to reduce everything to 2 dimensions
scaled_data = StandardScaler().fit_transform(numeric_features_clean)
pca_result = PCA(n_components=2, random_state=42).fit_transform(scaled_data)

plt.figure(figsize=(6, 5))

plt.scatter(pca_result[y == 0, 0], pca_result[y == 0, 1],
            s=6, alpha=0.3, label='Not vaccinated', color='#4477aa')

plt.scatter(pca_result[y == 1, 0], pca_result[y == 1, 1],
            s=6, alpha=0.3, label='Vaccinated', color='#cc6677')

plt.legend()
plt.title('PCA Projection (2 Components)')
plt.xlabel('Principal Component 1')
plt.ylabel('Principal Component 2')

# Display the plot
plt.show()
```

Preprocessing and feature engineering

```python
categorical_cols = ['age_group', 'education', 'race', 'sex', 'income_poverty',
                    'marital_status', 'rent_or_own', 'employment_status',
                    'census_msa', 'employment_sector']
test=pd.read_csv('dataset_C_testing.csv')
X = train.drop(columns=[target_col, 'respondent_id'], errors='ignore').copy()
X_test = test.drop(columns=['respondent_id'], errors='ignore').copy()


behavioral_cols = [
    'behavioral_antiviral_meds', 'behavioral_avoidance', 'behavioral_face_mask',
    'behavioral_wash_hands', 'behavioral_large_gatherings',
    'behavioral_outside_home', 'behavioral_touch_face'
]

# Apply feature engineering to both training and test sets
for dataset in (X, X_test):
    # Engineered Feature 1: Total sum of precautions taken
    dataset['total_precautions'] = dataset[behavioral_cols].sum(axis=1)

    # Engineered Feature 2: Interaction between doctor recommendation and perceived risk
    dataset['doc_x_risk'] = dataset['doctor_recc_covid'].fillna(0) * dataset['opinion_covid_risk'].fillna(0)

# Define numeric columns (everything not in the categorical list)
numeric_cols = [col for col in X.columns if col not in categorical_cols]

# Pipeline 1: Standard pre-processor (OneHot Encoding)
standard_preprocessor = ColumnTransformer(transformers=[
    ('num', Pipeline(steps=[
        ('imputer', SimpleImputer(strategy='median')),
        ('scaler', StandardScaler())
    ]), numeric_cols),
    ('cat', OneHotEncoder(handle_unknown='ignore'), categorical_cols)
])

print(f"Numeric features: {len(numeric_cols)} | Categorical features: {len(categorical_cols)}")

# Pipeline 2: OrdinalEncoder for Gradient Boosting
boosting_preprocessor = ColumnTransformer(transformers=[
    ('cat', OrdinalEncoder(handle_unknown='use_encoded_value', unknown_value=-1), categorical_cols),
    ('num', 'passthrough', numeric_cols)
])

boosting_cat_indices = list(range(len(categorical_cols)))
```

```python
def run_logistic_regression(X_train, y_train, X_test):
    print("Training Logistic Regression...")
    pipeline = Pipeline([
        ('pre', standard_preprocessor),
        ('m', LogisticRegression(max_iter=4000, C=0.3))
    ])

    # Get cross-validation predictions
    val_preds = cross_val_predict(pipeline, X_train, y_train, cv=kfold, method='predict_proba', n_jobs=5)[:, 1]

    # Fit on the whole dataset and predict on test set
    pipeline.fit(X_train, y_train)
    test_preds = pipeline.predict_proba(X_test)[:, 1]

    score = roc_auc_score(y_train, val_preds)
    print(f"Logistic Regression ROC-AUC: {score:.4f}")

    return pipeline, val_preds, test_preds

# Member 1 runs their function
trained_models['LogisticRegression'], val_predictions['LogisticRegression'], test_predictions['LogisticRegression'] = run_logistic_regression(X, y, X_test)
```
